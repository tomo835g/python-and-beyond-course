## Colab とは

**Google Colab**（コラブ）は、ブラウザ（Chrome や Safari など、ふだんインターネットを見るソフト）だけで **Python**（パイソン、プログラミング言語の一つ）を動かせる Google のサービスです。

- 自分のパソコンに何かをインストールする必要はありません。
- 使うには、個人の Google アカウントでログインします。
- 計算は Google のコンピュータが行うので、パソコンの性能はあまり関係ありません。

## 教材を開いて、コピーを保存する

この教材のいちばん上にある「Open in Colab」のボタンを押すと、Colab でこの教材が開きます。

ただし、開いただけの状態では、**あなたが書いたものは保存されません**。ページを閉じると消えてしまいます。次の手順で、自分の Google ドライブにコピーを保存しましょう。

1. 画面の左上あたりにある「ドライブにコピーを保存」というメニュー（または「ファイル」メニューの中の同じ名前の項目）を選びます。
2. 新しいタブでコピーが開きます。以後はそのコピーで作業します。

画面の見た目や項目の名前は変わることがあります。見つからないときは、「ファイル」メニューを開いて「コピー」「保存」という言葉を探してください。

## セルとは

Colab のノートブックは、**セル**という四角い箱が縦に並んだ形をしています。セルには2種類あります。

- **テキストセル**：いま読んでいるような、説明の文章を書くセルです。
- **コードセル**：Python のプログラム（**コード**）を書くセルです。左側に再生ボタン（▶）が付いています。

## セルを実行する

次のコードセルには `1 + 1` と書いてあります。セルをクリックして、次のどちらかをしてみましょう。

- 左側の再生ボタン（▶）を押す。
- キーボードで `Shift` を押しながら `Enter` を押す（Mac も Windows も同じ）。

セルの中身を Python が読んで動かすことを**実行**といいます。

初めて実行したとき、「このノートブックは Google が作成したものではありません」のような警告が出ることがあります。この教材を開いた場合は、そのまま実行してかまいません（「このまま実行」のようなボタンを押します。表示は変わることがあります）。

```python
1 + 1
```

セルの下に `2` と表示されたはずです。これが実行結果です。

セルの**最後の行の値は、自動で表示されます**。

**やってみよう**：`1 + 1` を `10 + 20` に書き換えて、もう一度実行してみましょう。

## 上から順に実行する

セルは**上から順に**実行するのが基本です。この回の後半や次の回以降では、前のセルで作ったものを後のセルが使うことがあります。途中を飛ばすと、エラー（うまく動かなかったという知らせ）が出ることがあります。

エラーが出ても壊れることはないので、安心してください。

## 困ったとき

- 動きがおかしくなったら、画面上の「ランタイム」というメニューから「ランタイムを再起動」のような項目を選ぶと、Python を最初の状態に戻せます。
- 最初から全部を順に動かし直したいときは、同じ「ランタイム」メニューの中の「すべてのセルを実行」のような項目を選びます。

メニューの名前は変わることがあります。「ランタイム」や「再起動」「すべて実行」といった言葉を探してください。見つからなければ、「Colab でランタイムを再起動するにはどうすればよいですか」のように AI に聞くのも一つの方法です（聞き方は第2回で説明します）。

## 計算してみよう

Python は電卓のように使えます。計算に使う記号は次のとおりです。

| 記号 | 意味 | 例 |
|---|---|---|
| `+` | たし算 | `3 + 2` |
| `-` | ひき算 | `3 - 2` |
| `*` | かけ算 | `3 * 2` |
| `/` | わり算 | `7 / 2` |
| `//` | わり算の商（整数部分） | `7 // 2` |
| `%` | わり算の余り | `7 % 2` |
| `**` | べき乗 | `2 ** 3` |

```python
7 / 2
```

```python
7 // 2
```

```python
7 % 2
```

```python
2 ** 10
```

`7 / 2` は `3.5` になります。割り算の結果は、割り切れるときも**小数**になります。たとえば `6 / 2` は `3.0` です。

```python
6 / 2
```

**やってみよう**：`17 // 5` と `17 % 5` を、それぞれ実行してみましょう。「17 を 5 で割ると、商が何で余りが何か」と合っていますか。

## かっこで計算の順番を決める

数学と同じで、かけ算・わり算はたし算・ひき算より先に計算されます。先に計算したいところは、かっこ `( )` で囲みます。

```python
2 + 3 * 4
```

```python
(2 + 3) * 4
```

**やってみよう**：`(2 + 3) * 4` のかっこの位置を変えて、答えがどう変わるか試してみましょう。

## print() で表示する

セルの最後の行の値は自動で表示されますが、途中の行の値は表示されません。

```python
1 + 1
2 + 2
3 + 3
```

表示されるのは最後の `6` だけです。いくつも表示したいときは、`print()` を使います。かっこの中に、表示したいものを入れます。

```python
print(1 + 1)
print(2 + 2)
print(3 + 3)
```

文字を表示したいときは、`"` で囲みます。

```python
print("こんにちは")
```

`"こんにちは"` のように `"` で囲んだものは、計算ではなく**文字**として扱われます。文字の詳しい話は第3回でします。いまは「文字は `"` で囲む」とだけ覚えておけば十分です。

`print()` の中にはコンマ区切りで複数のものを並べられます。

```python
print("合計は", 1 + 2, "です")
```

**やってみよう**：`"こんにちは"` を自分の名前に書き換えて実行してみましょう。

## コメント

`#`（シャープ）から行の終わりまでは、Python は無視します。これを**コメント**といいます。自分や他の人のためのメモを書くのに使います。

```python
# これはコメントです。実行されません
print(100 * 3)  # 行の途中からでも書けます
```

コードの右側に「何の計算か」を書いておくと、あとで読み返すときに助かります。

## 実例：買い物の合計と割り勘

友だち4人で、お菓子を買って割り勘にします。

```python
# 買ったものの値段（円）
snack = 280
juice = 150
pizza = 1200

# 合計
total = snack + juice * 3 + pizza
print("合計金額", total, "円")
```

ここで `snack = 280` のように書くと、`280` という数に `snack` という名前を付けられます。この名前を**変数**といいます（詳しくは第3回「変数と型」で説明します。いまは「数に名前を付けておける」と思っておけば十分です）。

次に、4人で割ります。

```python
people = 4

# 1人あたり（小数になる）
print("1人あたり", total / people, "円")

# 1人あたりの整数の金額と、余りを求める
print("1人あたり（整数）", total // people, "円")
print("余り", total % people, "円")
```

割り算 `/` では小数、`//` と `%` では「商」と「余り」が出ます。お金を集めるときは「1人 `//` の金額を集めて、余りは誰かが出す」といった使い方ができます。
