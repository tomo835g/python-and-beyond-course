## 2. 聞き方の基本

AI は、こちらが書いたことしか知りません。画面は見えていませんし、あなたが何を作ろうとしているかも知りません。ですから、質問には次の2つを入れます。

**何を貼るか**

- 分からないコード（そのままコピーして貼る）
- エラーが出たときは、エラーの文の**全文**（途中で切らない）

**何を書くか**

- 何をしたいのか（目的）
- 何を試したのか
- 自分がプログラミングの初心者であること

まず、悪い聞き方を見てみましょう。

```text
動きません。
```

これでは、AI は何が動かないのか分からず、「コードを見せてください」と聞き返すか、的外れな答えを返します。

次は良い聞き方です。

```text
Python 初心者です。画面に「こんにちは」と表示したくて、次のコードを実行しました。

print("こんにちは)

すると、次のエラーが出ました。

  File "<ipython-input-1-xxxx>", line 1
    print("こんにちは)
          ^
SyntaxError: unterminated string literal (detected at line 1)

原因と、直し方を、初心者にも分かるように教えてください。
```

違いは次の通りです。

| | 悪い聞き方 | 良い聞き方 |
|---|---|---|
| コード | なし | そのまま貼ってある |
| エラー | なし | 全文が貼ってある |
| 目的 | なし | 「こんにちは」と表示したい |
| 自分の状況 | なし | 初心者だと伝えている |

覚えておくこと：**コード、エラーの全文、目的、初心者であること**の4つを入れる。

## 3. 場面別の聞き方

そのまま写して使える形で、6つの場面を並べます。`〜` や `（ここに貼る）` の部分を、自分のものに置き換えてください。

### 場面1：コードの意味が分からない

コードを貼って、1行ずつ説明してもらいます。

```text
Python 初心者です。次のコードを1行ずつ説明してください。

（ここにコードを貼る）
```

### 場面2：エラーが出た

コードとエラーの全文を貼り、原因と直し方を聞きます。

```text
Python 初心者です。〜をしたくて次のコードを実行したら、エラーが出ました。

（ここにコードを貼る）

エラーの文は次の通りです。

（ここにエラーの全文を貼る）

原因と、直し方を教えてください。
```

### 場面3：用語が分からない

たとえ話で説明してもらうと、イメージがつかみやすくなります。

```text
Python 初心者です。「〜」という言葉の意味が分かりません。身近なたとえ話で説明してください。
```

### 場面4：例題が解けない

答えを聞いてしまうと身につかないので、ヒントだけをもらいます。

```text
Python 初心者です。次の問題の答えは言わずに、ヒントを1つだけください。

（ここに問題を貼る）
```

まだ分からなければ、「もう1つヒントをください」と続けます。

### 場面5：自分の答えが合っているか確かめたい

自分で書いたあとに、確認だけを頼みます。

```text
Python 初心者です。次の問題に、自分でこう答えました。この答えを確かめて、間違いがあれば理由を教えてください。直したコードは、私が自分で書き直すので、まだ見せないでください。

問題：（ここに問題を貼る）
私の答え：（ここにコードを貼る）
```

### 場面6：教材の説明が分かりにくい

教材の文章をそのまま貼って、別の言い方にしてもらいます。

```text
Python 初心者です。次の説明が分かりにくいので、別の言い方で説明してください。

（ここに教材の文章を貼る）
```

## 4. AI の答えを確かめる

AI はとても便利ですが、**間違えることがあります**。しかも、間違っていても自信ありげに書きます。ですから、次の習慣をつけてください。

- **必ず動かして確かめる**：AI が書いたコードは、Colab のセルに写して実行します。動いて、思った結果になるかを見ます。
- **小さく試す**：長いコードをまとめて貼るのではなく、1行ずつ、または短い部分ごとに動かします。
- **おかしいと思ったら聞き直す**：「実行したらこのエラーが出ました」と、結果を伝えて聞き直せば、AI は直してくれます。

AI がしがちな間違いには、次のようなものがあります。

- **古い書き方を教える**：昔は動いたが、今は動かない書き方。
- **存在しない命令をそれらしく書く**：本当はない名前の命令を、あるかのように書くことがあります。

実際に、存在しない命令を呼ぶとどうなるか見てみましょう。次のセルはわざとエラーが出ます。

````python
# 「say_hello」という命令は Python にはありません。実行するとエラーになります
say_hello("こんにちは")
````

`NameError: name 'say_hello' is not defined`（`say_hello という名前は定義されていません`）と出ます。AI が書いたコードにこのようなエラーが出たら、「その命令は本当にありますか？」と AI に聞き返しましょう。

## 5. 気をつけること

- **まず自分で考える**：例題の答えをすぐ AI に聞くと、読んだ気になるだけで身につきません。5分ほど自分で考え、試してから聞きましょう。聞くときも、まずは「ヒントを1つだけ」にします。
- **個人情報やパスワードを貼らない**：名前、学籍番号、住所、パスワード、メールアドレスなどを AI への質問に入れてはいけません。コードの中にそれらが入っていたら、消してから貼ります。
- **AI の答えをそのまま提出しない**：授業のレポートや課題の扱いは、授業のルールに従ってください。

## 7. 聞き方の早見表

| 場面 | 送る文の例 |
|---|---|
| コードの意味が分からない | 「Python 初心者です。次のコードを1行ずつ説明してください。」＋コード |
| エラーが出た | 「〜をしたくて実行したらエラーが出ました。原因と直し方を教えてください。」＋コード＋エラーの全文 |
| 用語が分からない | 「『〜』の意味を、身近なたとえ話で説明してください。」 |
| 例題が解けない | 「答えは言わずに、ヒントを1つだけください。」＋問題 |
| 自分の答えを確かめたい | 「この答えを確かめて、間違いがあれば理由を教えてください。」＋問題＋自分の答え |
| 教材の説明が分かりにくい | 「別の言い方で説明してください。」＋教材の文章 |

どの場面でも共通すること：コードとエラーは全文を貼る、目的を書く、初心者であると伝える、答えは必ず動かして確かめる、個人情報は貼らない。

## この回のポイント

- 質問には、**コード、エラーの全文、目的、初心者であること**を書く。
- AI は間違えることがある。**必ず動かして確かめ、小さく試す**。
- まず自分で考えてから聞く。例題は「ヒントを1つだけ」から。個人情報やパスワードは貼らない。

## 解答

### 練習1の解答

````python
print("こんにちは")
````

文字を囲む `"` は、始まりと終わりの2つが必要です。終わりの `"` を書き忘れたのが原因でした。

### 練習2の解答

````python
print(1 + 2)
````

Python は大文字と小文字を区別します。`print` はすべて小文字です。`Print` だと「そんな名前は知りません」（`NameError`）になります。

### 練習3の解答

````python
print("こんにちは")
````

`（` `）` は全角のかっこで、Python が使うのは半角の `(` `)` です。見た目は似ていても、別の文字として扱われます。同じ理由で、全角のスペースも、コードの中ではエラーのもとになります。コードは、半角で書きましょう。
