## 変数：値に名前を付けてしまう

第1回では、`1 + 1` や `7 / 2` のように数を直接書いて計算しました（第1回の実例で、変数を少しだけ使いました）。同じ数を何度も使うときや、あとで値を変えたくなったときは、値に**名前**を付けておくと便利です。名前を付けて値をしまっておく入れ物を**変数**と呼びます。

変数に値を入れるには `=` を使います。これを**代入**といいます。

```python
price = 300          # price（値段）という名前に 300 をしまう
print(price)         # 名前を書くと、しまった値が取り出せる
print(price * 2)     # 計算にも使える
```

大事な注意があります。Python の `=` は、数学の「等しい」ではありません。**「右側の値を、左側の名前にしまう」**という命令です。`price = 300` は「price は 300 に等しい」ではなく、「300 を price にしまえ」と読んでください。

## 上書きと「+=」

変数には、あとから別の値を入れ直せます。これを**上書き**といいます。古い値は消えて、新しい値に置き換わります。

```python
count = 1
print(count)
count = 5            # 上書き：1 は消えて 5 になる
print(count)
```

右側に、その変数自身を使うこともできます。

```python
x = 10
x = x + 1            # 右側を先に計算（10 + 1 = 11）、その結果を x にしまう
print(x)
```

数学では `x = x + 1` は成り立ちませんが、Python では「今の x に 1 を足して、x に入れ直す」という意味です。ここが `=` が「等しい」ではない、いちばん分かりやすい例です。

同じことは `+=` と短く書けます。`-=`、`*=`、`/=` もあります。

```python
x = 10
x += 1               # x = x + 1 と同じ
x += 1
print(x)
x *= 3               # x = x * 3 と同じ
print(x)
```

## 変数の名前の付け方

変数の名前は自由に付けられますが、守る決まりと、守ると読みやすくなる習慣があります。

守る決まり（守らないとエラーになります）：

- 数字で始めてはいけない（`1st` はエラー、`first` はよい）
- 空白や `-` は使えない（`total price` や `total-price` はだめ）
- `print` のように Python がすでに使っている名前は、使わない（使うと `print()` が使えなくなります）

読みやすくする習慣：

- 英語（半角英数字）で書く
- 小文字を使い、単語の区切りは `_`（アンダースコア）にする（例：`total_price`）
- 意味が分かる名前にする（`a` や `x` より `price` や `age`）

```python
unit_price = 120     # 単価
number_of_items = 3  # 個数
total_price = unit_price * number_of_items
print(total_price)
```

名前を見れば何の値か分かるので、コメントがなくても読めます。数日後の自分が読んでも困らないように、名前は丁寧に付けましょう。

## 型：値の種類

Python の値には**型**（かた）という種類があります。最初に覚えるのは次の4つです。

| 型 | 名前 | 例 |
| --- | --- | --- |
| `int` | 整数 | `3`、`-10`、`0` |
| `float` | 小数 | `3.14`、`-0.5` |
| `str` | 文字列（文字のならび） | `"こんにちは"`、`"abc"` |
| `bool` | 真偽値（正しいか正しくないか） | `True`、`False` |

値の型は `type()` で調べられます。

```python
print(type(3))
print(type(3.14))
print(type("こんにちは"))
print(type(True))
```

`<class 'int'>` のように表示されます。`int` や `float` の部分が型の名前です。

注意：`3` と `3.0` は、見た目は似ていますが型が違います（`3` は `int`、`3.0` は `float`）。また `/` で割り算をすると、割り切れても結果は `float` になります。

```python
print(type(3))
print(type(3.0))
print(type(6 / 2))   # 割り切れても float
print(6 / 2)
```

`True` と `False` は、頭の文字が大文字です。`"True"` のように `""` で囲むと、文字列になります。

```python
print(type(True))
print(type("True"))  # 引用符で囲むと文字列
```

変数に入れた値も、`type()` で調べられます。

```python
age = 19
print(type(age))
```

### やってみよう

## 文字列

文字列は `""`（ダブルクォーテーション）で囲みます。`''`（シングルクォーテーション）でも同じです。

```python
name1 = "田中"
name2 = '佐藤'
print(name1, name2)
```

使い分けのコツは、文字列の中に引用符を含めたいときです。外側と中身で種類を変えれば、そのまま書けます。

```python
print("It's a pen.")          # 中に ' を含む文字列は "" で囲む
print('彼は"はい"と言った')    # 中に " を含む文字列は '' で囲む
```

文字列も計算のようなことができます。`+` は**つなげる**、`*` は**くり返す**です。

```python
last_name = "山田"
first_name = "花子"
print(last_name + first_name)   # つなげる
print("ha" * 3)                 # 3回くり返す
print("-" * 20)                 # 区切り線を作るのによく使う
```

文字列の長さ（文字数）は `len()` で調べます。

```python
word = "Python"
print(len(word))
print(len("こんにちは"))
```

## 型の変換：int()、float()、str()

型がちがう値どうしは、そのままでは計算できないことがあります。たとえば、文字列の `"1"` と数の `1` を足してみましょう。次のセルは**わざとエラーを出します**。

```python
print("1" + 1)
```

`TypeError`（型のエラー）が出ます。メッセージを読むと、文字列と整数は足せない、という意味のことが書かれています。見た目は同じ「1」でも、`"1"` は「文字としての1」、`1` は「数としての1」なので、Python は足し方が決められないのです。

直し方は、型をそろえることです。型を変えるには、次の**関数**を使います。関数とは、`print()` や `type()` のように、名前の後のかっこの中に値を入れて使う命令のことです（詳しくは第8回「関数」で説明します）。

- `int(値)`：整数にする
- `float(値)`：小数にする
- `str(値)`：文字列にする

```python
print(int("1") + 1)      # "1" を整数にしてから足す → 数どうしの足し算で 2
print("1" + str(1))      # 1 を文字列にしてから足す → 文字のつなぎ合わせで "11"
```

どちらにそろえるかで結果が変わります。何をしたいのか（計算したいのか、文字をつなげたいのか）で選びます。

ほかの変換の例です。

```python
print(int(3.9))          # 小数 → 整数（小数点以下を捨てる）
print(float("2.5"))      # 文字列 → 小数
print(float(3))          # 整数 → 小数
print(str(100) + "円")   # 数 → 文字列にして文字とつなぐ
```

`int("abc")` のように、数として読めない文字列を変換しようとすると、やはりエラーになります。

## f 文字列：文字列の中に変数を埋め込む

文字列と数を組み合わせて文章を作るとき、`+` と `str()` を使うと書くのが大変です。

```python
name = "田中"
age = 19
print(name + "さんは" + str(age) + "歳です")
```

そこで便利なのが **f 文字列**です。文字列の前に `f` を付け、埋め込みたい変数を `{}` で囲みます。型の変換も自動でやってくれます。

```python
name = "田中"
age = 19
print(f"{name}さんは{age}歳です")
```

`{}` の中には、変数だけでなく計算式も書けます。

```python
print(f"来年は{age + 1}歳です")
```

小数の桁数を決めたいときは、`{値:.2f}` のように書きます（小数点以下2桁）。

```python
pi = 3.14159265
print(f"円周率は{pi:.2f}です")
```

## 比較：True と False を作る

数や文字列を比べると、答えとして `True`（正しい）か `False`（正しくない）の真偽値ができます。

| 書き方 | 意味 |
| --- | --- |
| `a == b` | a と b は等しい |
| `a != b` | a と b は等しくない |
| `a < b` | a は b より小さい |
| `a > b` | a は b より大きい |
| `a <= b` | a は b 以下（小さいか等しい） |
| `a >= b` | a は b 以上（大きいか等しい） |

`<=` と `>=` は、`<` や `>` の後に `=` を続けて書きます（`=<` の順ではエラーになります）。等しいかどうかを調べるときは `=` を2つ並べた `==` を使います。`=` は代入なので、混同しないようにしましょう。

```python
age = 19
print(age == 19)
print(age != 19)
print(age < 20)
print(age > 20)
```

```python
is_adult = age >= 18     # 結果の True/False も変数にしまえる
print(is_adult)
print(type(is_adult))
```

比較の結果を使って「もし〜ならこうする」と処理を分ける方法は、第6回「条件分岐」で詳しくやります。ここでは「比べると True か False ができる」ことだけ覚えておけば十分です。

### やってみよう

## 実例：レシートの1行を作る

ここまでの内容を組み合わせて、買い物のレシートの1行を作ってみましょう。商品の名前・値段・個数から、合計金額を計算して表示します。

```python
item_name = "ノート"
unit_price = 180
quantity = 3

subtotal = unit_price * quantity   # 小計
print(f"{item_name}　{unit_price}円 × {quantity}個 = {subtotal}円")
```

税込みの金額（税率10%）も出してみます。小数になるので `int()` で整数にします。

```python
tax_rate = 0.1
total = int(subtotal * (1 + tax_rate))
print(f"{item_name}の税込み合計：{total}円")
print("-" * 20)
```

文字列の `"…"` の中なら、全角の文字や全角スペース（`ノート` の後の空白）を使ってもかまいません。第2回で見たように、全角が困るのは `"…"` の外のコードの部分です。

商品や個数を変えれば、同じ式のままで別のレシートが作れます。これが変数を使う利点です。

### やってみよう

`item_name`、`unit_price`、`quantity` を変えて、レシートの行がどう変わるか確かめましょう。
